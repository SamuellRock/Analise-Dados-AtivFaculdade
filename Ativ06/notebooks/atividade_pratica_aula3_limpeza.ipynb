## 4. Batalha 1 — A tirania dos tipos de dados

Nos slides, vimos que datas não podem ser tratadas como texto e que números em formato string quebram análises. fileciteturn2file0

### Tarefa
Converta corretamente, quando fizer sentido:
- `data`
- `receita`
- `lucro`
- `quantidade`

### Orientação
- Investigue valores estranhos antes de converter
- Pense no uso de `errors='coerce'`
- Registre em markdown ou comentário quais problemas encontrou


```python
# Explore valores problemáticos e faça as conversões de tipo
def to_numeric_br(serie):
    serie_limpa = (
        serie.astype(str)
        .str.strip()
        .str.replace("R$", "", regex=False)
        .str.replace("r$", "", regex=False)
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
        .replace({"": np.nan, "Nan": np.nan, "None": np.nan})
    )
    return pd.to_numeric(serie_limpa, errors="coerce")

# difinindo o formato da data
print("Amostras de data:")
print(df["data"].astype(str).drop_duplicates().head(10))
print()

print("Amostras de receita:")
print(df["receita"].astype(str).drop_duplicates().head(10))
print()

if "lucro" in df.columns:
    print("Amostras de lucro:")
    print(df["lucro"].astype(str).drop_duplicates().head(10))
    print()

print("Amostras de quantidade:")
print(df["quantidade"].astype(str).drop_duplicates().head(10))



print(df.dtypes)
print()
print(df.head())

### Problemas encontrados
# A coluna `data` possui valores em formato textual.
# As colunas `receita` e `lucro` continham números como texto, com símbolo `R$` e separadores decimais/milhar.
# Foi necessário limpar esses caracteres antes da conversão.
# O uso de `errors='coerce'` evita erro de execução e transforma valores inválidos em `NaN`.
# A coluna `quantidade` foi convertida para tipo numérico, também com `errors='coerce'`, para tratar possíveis valores não numéricos.

### Colunas críticas identificadas
# `data`: contém valores ausentes (NaN)
# `receita`: valores incorretos após conversão e possíveis NaN
# `lucro`: mesma situação da receita
# `quantidade`: aparentemente consistente, mas requer validação

#### data
# remoção de linhas com valores nulos
# Justificativa: a data é essencial para análise temporal e não pode ser inferida com segurança

#### receita
# correção da conversão e remoção de valores nulos
# Justificativa: é uma variável central de negócio, não sendo adequado imputar valores

#### lucro
# correção da conversão e manutenção inicial de nulos
# Justificativa: pode ser mantido caso não seja foco da análise; remoção apenas se necessário

#### quantidade
# conversão para numérico e possível imputação com mediana
# Justificativa: variável discreta onde imputação é aceitável e menos impactante

print('\n\n--------------------------------------------------------------------------------------------------------------------')
```

## 5. Batalha 2 — O enigma dos valores ausentes

A aula reforça que valores ausentes não devem ser tratados automaticamente; a decisão precisa ser justificada. fileciteturn2file0

### Tarefa
1. Mapeie os valores ausentes por coluna
2. Identifique quais colunas críticas têm nulos
3. Defina uma estratégia para cada caso:
   - remover linhas?
   - preencher?
   - manter?
4. Justifique cada escolha

### Dica
Evite decisões automáticas sem análise de contexto.


```python
# Investigue os nulos e aplique aqui sua estratégia de tratamento
nulos_antes = df.isna().sum()

colunas_criticas = [col for col in ["data", "receita", "uf", "canal", "produto"] if col in df.columns]
df_tratado_nulos = df.dropna(subset=colunas_criticas).copy()

if "categoria" in df_tratado_nulos.columns:
    df_tratado_nulos["categoria"] = df_tratado_nulos["categoria"].fillna("Não informado")

for col in ["quantidade", "custo_unitario", "preco_unitario"]:
    if col in df_tratado_nulos.columns:
        mediana_col = df_tratado_nulos[col].median()
        df_tratado_nulos[col] = df_tratado_nulos[col].fillna(mediana_col)

nulos_depois = df_tratado_nulos.isna().sum()

print("Nulos antes:")
print(nulos_antes)
print()
print("Nulos depois:")
print(nulos_depois)

### Tratamento de valores nulos

#### 1. Identificação
# As colunas com valores nulos identificadas foram:
    # `data` (224 nulos)
    # `receita` (5 nulos)
    # `uf` (5 nulos)
    # `canal` (5 nulos)
    # `categoria` (5 nulos)
    # `lucro` (6 nulos)
    # `observacao` (50 nulos)

#### 2. Estratégia adotada

# Colunas críticas (`data`, `receita`, `uf`, `canal`):
# Estratégia: remoção das linhas com valores nulos (`dropna`)
# Justificativa:
  # `data` é essencial para análises temporais
  # `receita` é uma métrica principal
  # `uf` e `canal` são importantes para segmentação
# Observação:
  # A remoção de 224 registros devido à ausência de `data` pode impactar o volume de dados disponível

# Coluna `categoria`:
# Estratégia: preenchimento com "Não informado"
# Justificativa:
  # Não é uma coluna crítica
  # Mantém os dados para análise sem perda significativa

# Colunas numéricas (`quantidade`, `custo_unitario`, `preco_unitario`):
# Estratégia: preenchimento com a mediana
# Justificativa:
  # A mediana é menos sensível a valores extremos (outliers)
  # Evita distorção nos dados

# Coluna `observacao`:
# Estratégia: manter valores nulos
# Justificativa:
  # Campo descritivo sem impacto direto nas análises quantitativas


print('\n\n--------------------------------------------------------------------------------------------------------------------')
```

```python
# Investigue duplicidades e trate o problema conscientemente
duplicados_antes = df_tratado_nulos.duplicated().sum()

df_sem_duplicados = df_tratado_nulos.drop_duplicates().copy()

duplicados_depois = df_sem_duplicados.duplicated().sum()

print("Duplicados antes:", duplicados_antes)
print("Duplicados depois:", duplicados_depois)
```

## 7. Padronização de categorias

Os slides mostram que categorias duplicadas ou mal escritas podem distorcer rankings e filtros. fileciteturn2file0

### Tarefa
Inspecione e padronize, se necessário:
- `uf`
- `canal`
- `categoria`

### Pense em:
- maiúsculas e minúsculas
- espaços extras
- acentuação / variações
- categorias equivalentes escritas de formas diferentes


```python
# Faça aqui a padronização de categorias
df_padronizado = df_sem_duplicados.copy()

if "uf" in df_padronizado.columns:
    df_padronizado["uf"] = df_padronizado["uf"].astype(str).str.strip().str.upper()

if "canal" in df_padronizado.columns:
    canal_base = df_padronizado["canal"].astype(str).str.strip().str.lower()
    df_padronizado["canal"] = canal_base.replace({
        "e-commerce": "E-commerce",
        "e commerce": "E-commerce",
        "ecommerce": "E-commerce",
        "loja fisica": "Loja Física",
        "loja física": "Loja Física",
        "marketplace": "Marketplace",
        "online": "Online"
    })

if "categoria" in df_padronizado.columns:
    df_padronizado["categoria"] = df_padronizado["categoria"].astype(str).str.strip()

if "produto" in df_padronizado.columns:
    df_padronizado["produto"] = df_padronizado["produto"].astype(str).str.strip()

if "canal" in df_padronizado.columns:
    canais_padronizados = df_padronizado["canal"].value_counts(dropna=False)
    print(canais_padronizados)

if "uf" in df_padronizado.columns:
    ufs_padronizadas = df_padronizado["uf"].value_counts(dropna=False)
    print()
    print(ufs_padronizadas)
```

## 8. Subindo de nível — Criação de variáveis derivadas

Depois da limpeza, é hora de enriquecer a base com variáveis úteis para análise. fileciteturn2file0

### Tarefa
Crie, se possível:
- `ano`
- `mes`
- `ano_mes`
- `margem_lucro`

### Atenção
A criação de `margem_lucro` exige cuidado com divisões por zero e possíveis valores infinitos.


```python
# Crie aqui as variáveis derivadas
df_enriquecido = df_padronizado.copy()

df_enriquecido["ano"] = df_enriquecido["data"].dt.year
df_enriquecido["mes"] = df_enriquecido["data"].dt.to_period("M").astype(str)

if "lucro" in df_enriquecido.columns:
    df_enriquecido["margem_lucro"] = np.where(
        df_enriquecido["receita"] > 0,
        df_enriquecido["lucro"] / df_enriquecido["receita"],
        np.nan
    )

print(df_enriquecido.head())
```
