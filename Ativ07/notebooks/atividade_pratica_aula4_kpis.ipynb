## 5. Métricas derivadas

A aula mostra que **somar valores não basta**.  
Precisamos criar métricas derivadas para avaliar eficiência e comportamento. fileciteturn5file0

### Tarefa
Crie, quando fizer sentido:
- uma coluna de **margem_lucro** (`lucro / receita`)

### Reflexão

A operação apresenta uma receita total que mostra o volume financeiro movimentado pelo negócio. O lucro total indica quanto desse volume realmente se converte em resultado após os custos. A margem de lucro ajuda a avaliar a eficiência da operação, mostrando se o faturamento está sendo bem capturado em rentabilidade. Já o ticket médio revela se o padrão de compra tende a valores mais altos ou mais baixos por venda.


```python
# Crie aqui as métricas derivadas
receita_total = df["receita"].sum()
lucro_total = df["lucro"].sum()
margem_lucro = lucro_total / receita_total
ticket_medio = df["receita"].mean()
margem_porc = margem_lucro * 100

print("Receita: ", receita_total)
print("Lucro: ", lucro_total)
print(f'Margem:  {margem_porc.round(2)}%')
```

## 6. Fatiando a realidade: análise por dimensão

A aula mostra que um KPI isolado não decide nada.  
É preciso perguntar: **como isso varia por canal, categoria, região e tempo?** fileciteturn5file0

### Tarefa
Liste, em markdown, quais colunas do dataset representam:
- métrica
- dimensão
- tempo


- Métricas: receita, lucro, margem_lucro
- Dimensões: canal_venda, categoria, uf, produto
- Tempo: data_venda


```python
df_kpi = df.copy()

df_kpi["margem_lucro"] = np.where(
    df_kpi["receita"] > 0,
    df_kpi["lucro"] / df_kpi["receita"],
    np.nan
)

ticket_medio_operacao = df_kpi["receita"].mean()

print(df_kpi[["receita", "lucro", "margem_lucro"]].head())
print()
print("Ticket médio da operação:", round(ticket_medio_operacao, 2))
```

## 7. Análise por canal

### Pergunta de negócio
**Qual canal de venda é o mais eficiente?**

### Tarefa
Crie uma tabela agregada por `canal_venda` contendo, no mínimo:
- Receita Total
- Lucro Total
- Margem Média ou Margem Calculada
- Quantidade de pedidos
- Ticket Médio

### Depois responda
1. Qual canal gera mais receita?
2. Qual canal gera maior eficiência?
3. Se você fosse gestor, priorizaria volume ou margem?


```python
# Gere aqui a tabela agregada por canal_venda
analise_canal = (
    df_kpi.groupby("canal_venda")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum"),
        quantidade_pedidos=("receita", "count"),
        ticket_medio=("receita", "mean")
    )
    .reset_index()
)

analise_canal["margem"] = np.where(
    analise_canal["receita_total"] > 0,
    analise_canal["lucro_total"] / analise_canal["receita_total"],
    np.nan
)

analise_canal = analise_canal.sort_values("receita_total", ascending=False)

print(analise_canal)    
```

## 8. Análise por categoria

### Pergunta de negócio
**Qual categoria é mais lucrativa e qual parece mais eficiente?**

### Tarefa
Crie uma tabela agregada por `categoria` com:
- Receita Total
- Lucro Total
- Margem
- Ticket Médio

### Depois responda
1. A categoria de maior receita também é a de maior margem?
2. Existe alguma categoria que parece vender bem, mas ser pouco eficiente?


```python
# Gere aqui a tabela agregada por categoria
analise_categoria = (
    df_kpi.groupby("categoria")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum"),
        ticket_medio=("receita", "mean")
    )
    .reset_index()
)

analise_categoria["margem"] = np.where(
    analise_categoria["receita_total"] > 0,
    analise_categoria["lucro_total"] / analise_categoria["receita_total"],
    np.nan
)

analise_categoria = analise_categoria.sort_values("lucro_total", ascending=False)

print(analise_categoria)
```

## 9. Análise por região (UF)

### Pergunta de negócio
**Quais UFs concentram a operação?**

### Tarefa
Monte uma tabela por `uf` com:
- Receita Total
- Lucro Total
- Participação percentual da receita
- Ticket Médio

### Depois responda
1. Quais UFs concentram maior receita?
2. Existe concentração regional?
3. Se fosse necessário expandir ou reforçar logística, por onde você começaria?


```python
# Gere aqui a análise por UF
analise_uf = (
    df_kpi.groupby("uf")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum"),
        ticket_medio=("receita", "mean")
    )
    .reset_index()
)

receita_total_uf = analise_uf["receita_total"].sum()

analise_uf["participacao_receita"] = analise_uf["receita_total"] / receita_total_uf
analise_uf = analise_uf.sort_values("receita_total", ascending=False)

print(analise_uf)
```

## 10. O motor da análise: agregações (groupby)

Nos slides, a lógica central é:
- `receita -> sum`
- `margem -> mean` ou cálculo derivado
- `produto -> count`
fileciteturn5file0

### Tarefa
Escolha **duas dimensões diferentes** e construa tabelas agregadas mostrando que você entendeu essa lógica.

Exemplos:
- por canal
- por categoria
- por UF
- por mês


```python
# Demonstre aqui o uso de groupby com diferentes agregações
agregacao_por_canal = (
    df_kpi.groupby("canal_venda")
    .agg(
        receita_total=("receita", "sum"),
        margem_media=("margem_lucro", "mean"),
        quantidade_pedidos=("produto", "count")
    )
    .reset_index()
    .sort_values("receita_total", ascending=False)
)

agregacao_por_uf = (
    df_kpi.groupby("uf")
    .agg(
        receita_total=("receita", "sum"),
        margem_media=("margem_lucro", "mean"),
        quantidade_pedidos=("produto", "count")
    )
    .reset_index()
    .sort_values("receita_total", ascending=False)
)

print("Agregação por canal:")
print(agregacao_por_canal)
print()
print("Agregação por UF:")
print(agregacao_por_uf)
```

## 11. Campeões e vilões (Top N)

A aula destaca o poder dos rankings curtos e focados. fileciteturn5file0

### Tarefa
Crie:
- Top 10 produtos por **lucro**
- Top 10 produtos por **receita**
- 5 categorias ou UFs com **pior margem** ou pior desempenho

### Perguntas
1. Quem são os “campeões”?
2. Quem são os “vilões”?
3. Que ação tática poderia ser tomada com base nisso?


```python
# Monte aqui os rankings Top N
top10_produtos_lucro = (
    df_kpi.groupby("produto")
    .agg(lucro_total=("lucro", "sum"))
    .reset_index()
    .sort_values("lucro_total", ascending=False)
    .head(10)
)

top10_produtos_receita = (
    df_kpi.groupby("produto")
    .agg(receita_total=("receita", "sum"))
    .reset_index()
    .sort_values("receita_total", ascending=False)
    .head(10)
)

piores_categorias_margem = (
    df_kpi.groupby("categoria")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum")
    )
    .reset_index()
)

piores_categorias_margem["margem"] = np.where(
    piores_categorias_margem["receita_total"] > 0,
    piores_categorias_margem["lucro_total"] / piores_categorias_margem["receita_total"],
    np.nan
)

piores_categorias_margem = piores_categorias_margem.sort_values("margem", ascending=True).head(5)

print("Top 10 produtos por lucro:")
print(top10_produtos_lucro)
print()
print("Top 10 produtos por receita:")
print(top10_produtos_receita)
print()
print("5 categorias com pior margem:")
print(piores_categorias_margem)
```

## 12. Análise temporal

A aula reforça que sem tempo a análise fica estática. fileciteturn5file0

### Tarefa
1. Converta `data_venda` para data, se necessário
2. Crie colunas auxiliares como:
   - mês
   - trimestre (opcional)
3. Gere uma tabela agregada por mês com:
   - receita
   - lucro
   - ticket médio ou margem

### Depois responda
1. O lucro parece crescer ao longo do tempo?
2. O último trimestre foi melhor que o anterior?
3. Quais meses concentram as vendas de fim de ano?


```python
# Faça aqui a análise temporal mensal
df_tempo = df_kpi.copy()

df_tempo["data_venda"] = pd.to_datetime(df_tempo["data_venda"], errors="coerce")
df_tempo["mes"] = df_tempo["data_venda"].dt.to_period("M")
df_tempo["trimestre"] = df_tempo["data_venda"].dt.to_period("Q")

analise_mensal = (
    df_tempo.groupby("mes")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum"),
        ticket_medio=("receita", "mean")
    )
    .reset_index()
)

analise_mensal["margem"] = np.where(
    analise_mensal["receita_total"] > 0,
    analise_mensal["lucro_total"] / analise_mensal["receita_total"],
    np.nan
)

analise_trimestral = (
    df_tempo.groupby("trimestre")
    .agg(
        receita_total=("receita", "sum"),
        lucro_total=("lucro", "sum")
    )
    .reset_index()
)

print("Análise mensal:")
print(analise_mensal)
print()
print("Análise trimestral:")
print(analise_trimestral)
```
